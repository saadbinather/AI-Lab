```python
L=[(1, 7), (1, 3), (3, 4, 5), (2, 2)]


max=0
for i in L:
    if max<(i[-1]):
       
        max=i[-1]

 

nL=[]
for i in range(0,max+1):
    for j in L:
         if j[-1]==i:
             nL.append(j)



print(nL)
    

```
