```python
def isEven(a):
    if a%2==0:
        return True
    else:
        return False

def Factorial(a):
    result=a
    num=a-1
    while num>0:
        result=result*num
        num-=1
    return result

print (isEven(3))
print(Factorial(4))
```

```python
str=input("Enter a string: ")
str=list(str)

i=0
j=len(str)-1
while i<j:
    temp=str[i]
    str[i]=str[j]
    str[j]=temp
    i+=1
    j-=1
print(str)
```

```python
s = input("enter string ")
left = 0
right = len(s) - 1
flag =0
while left < right:
    if s[left] != s[right]:
        flag = 1
        break
    left += 1
    right -= 1

print(flag==0)
```
