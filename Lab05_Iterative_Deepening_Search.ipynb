```python
from collections import deque
import heapq
```

```python
graph={

    'S':['A','B','C'],
    'A':['S','D','E'],
    'B':['S','F'],
    'C':['S','G'],
    'D':['A'],
    'E':['A'],
    'F':['B','H'],
    'G':['C'],
    'H':['F','T'],
    'T':['H'],
}
```

```python
def bfs(graph,start,goal,depth):
    level=0
    visited=set()
    visited.add(start)
    q=deque()
    q.append(start)
    l=[]
    while True:
        print('level ',level,'r esult')
        print(l)
        
       
        a=q.pop()
        l.append(a)

       
        

        level=level+1
        if level==depth:
            return False
        for neighbors in graph[a]:
            if(neighbors not in visited):
                visited.add(neighbors)
                q.append(neighbors)
                l.append(neighbors)
            if(neighbors==goal):
                return True 
    
```

```python
def ids(graph,start,goal):
    for i in range(0,len(graph)):
        bfs(graph,start,goal,i)
```

```python
bfs(graph,'S','T',0)
```
