```python
import numpy as np

# inputs
x = np.array([1.0, 2.0])

# Weights 
W = np.array([[0.5, -0.2],
              [0.3,  0.8]])


b = np.array([0.1, -0.1])

# ReLU 
def relu(x):
    return np.maximum(0, x)

# fwd
z = np.dot(x, W) + b
print("Weighted sum (hidden):", z)
h = relu(z)
print("Hidden layer output:", h)


Wo = np.array([0.1, 0.2])   






z_out = np.dot(h, Wo) 
print("Weighted sum (output):", z_out)



def softmax(x):
    exp_x = np.exp(x - np.max(x))  
    return exp_x / np.sum(exp_x)



output = softmax(np.array([z_out]))
print("Final output (softmax):", output)
```

```python

def sigmoid(x):
    return 1/np.exp(-x)

    
    
    
output = sigmoid(np.array([z_out]))
print("Final output (sigmoid):", output)
```
