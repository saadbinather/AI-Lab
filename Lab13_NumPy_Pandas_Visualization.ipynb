```python
#q4
df=pd.read_csv('income.csv')


print(df['age'].isna())
print(df['income'].isna())

mean=df['age'].mean()

df['newincome'] = np.where(df['income'] == '<=50K', 51, 49)
median=df['newincome'].median()



df['age'] = df['age'].fillna(mean)  # Replace NaN with 0
df['income'] = df['income'].fillna(median)  # Replace NaN with 0



print(df.head())
```

```python
#Q7
df=pd.read_csv('healthcare_dataset.csv')


print(df.head())





df = df.dropna(subset=['Age'])
df=df = df.dropna(subset=['Billing Amount'])


df['Date of Admission'] = pd.to_datetime(df['Date of Admission'], format='mixed')
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'], format='mixed')


df['Hospital_Stay_Days'] = (df['Discharge Date'] - df['Date of Admission']).dt.days

df['Hospital_Stay_Days'] = df['Hospital_Stay_Days'].fillna(0).astype(int)

print(df[['Date of Admission', 'Discharge Date', 'Hospital_Stay_Days']].head())

avg_stay = df['Hospital_Stay_Days'].mean()
avg_cost_per_diag = df.groupby('Medical Condition')['Billing Amount'].mean()



print(f"\nAverage Hospital Stay: {avg_stay:.2f} days")
print("\nAverage Treatment Cost per Diagnosis Type:")
print(avg_cost_per_diag)


Q1 = df['Billing Amount'].quantile(0.25)
Q3 = df['Billing Amount'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Billing Amount'] < lower_bound) | (df['Billing Amount'] > upper_bound)]
print(f"\nNumber of outliers detected in Treatment Cost: {len(outliers)}")


plt.figure(figsize=(12, 5))


plt.subplot(1, 2, 1)
avg_cost_per_diag.plot(kind='bar', color='skyblue')
plt.title('Avg Cost per Diagnosis')
plt.ylabel('Cost')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
plt.hist(df['Hospital_Stay_Days'], bins=15, color='salmon', edgecolor='black')
plt.title('Distribution of Hospital Stay')
plt.xlabel('Days')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```
