```python
import math

# leaf values
tree = [
    [3, 5, 6], # first MIN node
    [9, 1, 2] # second MIN node
]

pruned = []

def minimax(depth, node_index, maximizing_player, alpha, beta):

    # if we reached leaf level
    if depth == 2:
        return tree[node_index // 3][node_index % 3]

    if maximizing_player:
        best = -math.inf

        for i in range(2): # two MIN children
            val = minimax(depth + 1, i * 3, False, alpha, beta)
            best = max(best, val)
            alpha = max(alpha, best)

            if beta <= alpha:
                pruned.append("Subtree at MIN node " + str(i) + " pruned")
                break

        return best

    else: # MIN player
        best = math.inf

        for i in range(3): # three leaf nodes
            val = tree[node_index // 3][i]
            best = min(best, val)
            beta = min(beta, best)

            if beta <= alpha:
                pruned.append("Leaf " + str(tree[node_index // 3][i+1:]) + " pruned")
                break

        return best


optimal_value = minimax(0, 0, True, -math.inf, math.inf)

print("Optimal value at root:", optimal_value)

print("Pruned nodes:")
for p in pruned:
    print(p)
```

```python
import math

board = [
    [' ', ' ', ' '],
    [' ', ' ', ' '],
    [' ', ' ', ' ']
]

# print board
def print_board():
    for row in board:
        print(row[0], "|", row[1], "|", row[2])
        print("---------")

# check winner
def check_winner(player):

    # rows
    for i in range(3):
        if board[i][0] == player and board[i][1] == player and board[i][2] == player:
            return True

    # columns
    for i in range(3):
        if board[0][i] == player and board[1][i] == player and board[2][i] == player:
            return True

    # diagonals
    if board[0][0] == player and board[1][1] == player and board[2][2] == player:
        return True

    if board[0][2] == player and board[1][1] == player and board[2][0] == player:
        return True

    return False


def is_moves_left():
    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ':
                return True
    return False


# minimax with alpha beta pruning
def minimax(depth, is_max, alpha, beta):

    if check_winner('O'):
        return 10

    if check_winner('X'):
        return -10

    if not is_moves_left():
        return 0

    if is_max: # AI move
        best = -math.inf

        for i in range(3):
            for j in range(3):

                if board[i][j] == ' ':
                    board[i][j] = 'O'

                    value = minimax(depth+1, False, alpha, beta)

                    board[i][j] = ' '

                    best = max(best, value)
                    alpha = max(alpha, best)

                    if beta <= alpha:
                        break

        return best

    else: # Human move
        best = math.inf

        for i in range(3):
            for j in range(3):

                if board[i][j] == ' ':
                    board[i][j] = 'X'

                    value = minimax(depth+1, True, alpha, beta)

                    board[i][j] = ' '

                    best = min(best, value)
                    beta = min(beta, best)

                    if beta <= alpha:
                        break

        return best


# find best move for AI
def best_move():

    best_val = -math.inf
    move = (-1, -1)

    for i in range(3):
        for j in range(3):

            if board[i][j] == ' ':
                board[i][j] = 'O'

                move_val = minimax(0, False, -math.inf, math.inf)

                board[i][j] = ' '

                if move_val > best_val:
                    move = (i, j)
                    best_val = move_val

    board[move[0]][move[1]] = 'O'


# main game loop
while True:

    print_board()

    # human move
    row = int(input("Enter row (0-2): "))
    col = int(input("Enter col (0-2): "))

    if board[row][col] != ' ':
        print("Invalid move")
        continue

    board[row][col] = 'X'

    if check_winner('X'):
        print_board()
        print("Human Wins!")
        break

    if not is_moves_left():
        print_board()
        print("Draw!")
        break

    # AI move
    best_move()

    if check_winner('O'):
        print_board()
        print("AI Wins!")
        break

    if not is_moves_left():
        print_board()
        print("Draw!")
        break
```

```python
import random

# heuristic function: number of attacking pairs
def heuristic(state):

    attacks = 0

    for i in range(8):
        for j in range(i+1, 8):

            # same row
            if state[i] == state[j]:
                attacks += 1

            # diagonal attack
            if abs(state[i] - state[j]) == abs(i - j):
                attacks += 1

    return attacks


# generate neighbors
def get_neighbors(state):

    neighbors = []

    for col in range(8):

        for row in range(1,9):

            if row != state[col]:

                new_state = state.copy()
                new_state[col] = row
                neighbors.append(new_state)

    return neighbors


# hill climbing algorithm
def hill_climbing():

    # random initial state
    current = [random.randint(1,8) for i in range(8)]

    while True:

        current_h = heuristic(current)

        if current_h == 0:
            return current

        neighbors = get_neighbors(current)

        next_state = current
        next_h = current_h

        for n in neighbors:

            h = heuristic(n)

            if h < next_h:
                next_state = n
                next_h = h

        # if no better neighbor
        if next_h >= current_h:
            return current

        current = next_state


# run algorithm
solution = hill_climbing()

print("Solution State:", solution)
print("Heuristic Value:", heuristic(solution))
```
