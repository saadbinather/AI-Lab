```python
L=[1,2,3,4,5,6,99]

sq=lambda x: x*x
cb=lambda x:sq(x)*x


for i in L:
    print(f"{i}{" - "}{sq(i)}{" - "}{cb(i)}")
    

```

```python
L=[-4, 3, -2, 7, 5]

fL=list(filter(lambda num:(num>0),L))



nL=list(map(lambda num:num*num,fL))
print(nL)
```
